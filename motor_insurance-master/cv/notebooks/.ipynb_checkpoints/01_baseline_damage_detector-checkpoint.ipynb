# Module 0/2 POC: Baseline Vehicle Damage Detector

This notebook trains a first-pass YOLOv8 damage-detection model as a proof of concept, using a public car-damage dataset (Roboflow) as a stand-in until real Old Mutual claim photos are available.

**Scope**: Tier 1 from the project plan — basic damage detection (dent/scratch/crack/glass/etc.), not yet impact-zone classification or ordinal crush depth (those need the `vehicle_registry.py` mapping and are a follow-up notebook).

**Hardware note**: this machine has no dedicated GPU (Intel Iris Xe only). Training runs on CPU here — fine for a quick POC on a small subset/few epochs, but slow. For a real training run, export this notebook's cells into Google Colab (free T4 GPU) or point `DEVICE` at a CUDA box if you get access to one.

**Kernel**: select "Motor CV (POC)" (maps to `C:\mi_cv\venv`).

## 1. Environment check

```python
import torch, ultralytics, cv2, sys

print("python:", sys.version)
print("torch:", torch.__version__, "| cuda available:", torch.cuda.is_available())
print("ultralytics:", ultralytics.__version__)
print("opencv:", cv2.__version__)

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Training device:", DEVICE)
```

## 2. Dataset download (Roboflow)

Get a free API key at https://app.roboflow.com/ -> Settings -> Roboflow API, then put it in `cv/.env` as `ROBOFLOW_API_KEY=...` (copy `cv/.env.example`).

Default project below is the CarDD-style mirror (~2.5k images, 19 damage classes). Swap `workspace`/`project`/`version` for a different Roboflow Universe dataset if you prefer — e.g. the larger 8.9k-image Skillfactory set.

```python
from pathlib import Path
from dotenv import load_dotenv
import os

CV_ROOT = Path.cwd().parent  # cv/
load_dotenv(CV_ROOT / ".env")

ROBOFLOW_API_KEY = os.getenv("ROBOFLOW_API_KEY")
assert ROBOFLOW_API_KEY, "Set ROBOFLOW_API_KEY in cv/.env first (copy cv/.env.example)."

DATA_DIR = CV_ROOT / "data"
DATA_DIR.mkdir(exist_ok=True)
```

```python
from roboflow import Roboflow

rf = Roboflow(api_key=ROBOFLOW_API_KEY)

# CarDD-style mirror on Roboflow Universe: 2000 images (1395 train / 401 valid / 204 test), object-detection.
workspace = "car-damage-detection-cardd"
project_slug = "car-damage-severity-detection-cardd"
version = 3

project = rf.workspace(workspace).project(project_slug)
dataset = project.version(version).download("yolov8", location=str(DATA_DIR / project_slug))

print("Downloaded to:", dataset.location)
```

## 3. Quick EDA

Sanity-check class balance and look at a few annotated samples before spending CPU time training.

```python
import yaml

data_yaml_path = Path(dataset.location) / "data.yaml"
with open(data_yaml_path) as f:
    data_yaml = yaml.safe_load(f)

print("Classes:", data_yaml["names"])
print("Num classes:", data_yaml["nc"])

train_images = list((Path(dataset.location) / "train" / "images").glob("*"))
val_images = list((Path(dataset.location) / "valid" / "images").glob("*"))
print(f"Train images: {len(train_images)} | Val images: {len(val_images)}")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import random

def show_sample(img_path, ax):
    label_path = Path(str(img_path).replace("images", "labels")).with_suffix(".txt")
    img = plt.imread(img_path)
    h, w = img.shape[:2]
    ax.imshow(img)
    if label_path.exists():
        for line in label_path.read_text().splitlines():
            cls, cx, cy, bw, bh = map(float, line.split())
            x0, y0 = (cx - bw / 2) * w, (cy - bh / 2) * h
            rect = patches.Rectangle((x0, y0), bw * w, bh * h, linewidth=2, edgecolor="lime", facecolor="none")
            ax.add_patch(rect)
            ax.text(x0, y0 - 5, data_yaml["names"][int(cls)], color="lime", fontsize=8)
    ax.axis("off")

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
for ax, img_path in zip(axes.flat, random.sample(train_images, min(6, len(train_images)))):
    show_sample(img_path, ax)
plt.tight_layout()
plt.show()
```

## 4. Baseline training (YOLOv8n)

Nano model + few epochs so a CPU-only run finishes in a reasonable time. This is a POC baseline, **not** a production model — expect mediocre mAP. Bump `EPOCHS`/`IMGSZ`/model size once you move this to a GPU (Colab or otherwise).

```python
from ultralytics import YOLO

EPOCHS = 20       # bump to 100+ on GPU
IMGSZ = 640
BATCH = 8         # lower if you run out of RAM on CPU

model = YOLO("yolov8n.pt")  # COCO-pretrained nano checkpoint, auto-downloads once

results = model.train(
    data=str(data_yaml_path),
    epochs=EPOCHS,
    imgsz=IMGSZ,
    batch=BATCH,
    device=DEVICE,
    project=str(CV_ROOT / "models" / "runs"),
    name="baseline_yolov8n_damage",
)
```

## 5. Validate & visualize predictions

```python
metrics = model.val()
print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
```

```python
sample_val = random.sample(val_images, min(4, len(val_images)))
pred_results = model.predict(source=sample_val, conf=0.25, device=DEVICE)

fig, axes = plt.subplots(1, len(pred_results), figsize=(6 * len(pred_results), 6))
if len(pred_results) == 1:
    axes = [axes]
for ax, r in zip(axes, pred_results):
    ax.imshow(r.plot()[:, :, ::-1])
    ax.axis("off")
plt.tight_layout()
plt.show()
```

## 6. Export to ONNX

Matches the plan's production-inference decision: lightweight ONNX runtime in `damage_cv_model.py`, no training stack needed in the FastAPI service.

```python
onnx_path = model.export(format="onnx", imgsz=IMGSZ)
print("Exported:", onnx_path)
```
